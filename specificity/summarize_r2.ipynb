# Merges CMS, pMHC fold specificity, ProteinMPNN alanine scan, and AF2 ipSAE scores for all 199 designs and produces a ranked summary table.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
# Set pandas to display all columns
pd.set_option('display.max_columns', None)
# To also see all rows (optional)
pd.set_option('display.max_rows', None)
```

```python
CMS_PATH    = "/n/groups/marks/users/aaron/pmhc/specificity/CMS/outputs/r2/cms_scores.tsv"
PMHC_PATH   = "/n/groups/marks/users/aaron/pmhc/specificity/pMHC_fold/outputs/r2/analysis/pmhc_fold_specificity_scores.tsv"
MPNN_PATH   = "/n/groups/marks/users/aaron/pmhc/specificity/ProteinMPNN_ala/outputs/zero_shot_predictions/pepscan_kras-num_seq_per_target=10-use_mt_structure=0.csv"
AF2_PATH    = "/n/groups/marks/users/aaron/pmhc/af_init_guess/outputs/r2/af2_ipae_ipsae_scores.csv"
OUT_DIR     = "/n/groups/marks/users/aaron/pmhc/specificity/analysis/r2"
os.makedirs(OUT_DIR, exist_ok=True)

PEPTIDE_SEQ = "VVGADGVGK"  # p1-p9, D at p5
# Upward-facing positions (1-indexed in peptide): p5=D, p7=V, p9=K
# In the 8-row MPNN output (p4=A skipped), these are rows 4, 6, 8 (1-indexed within group)
# Confirmed by: V(p1) V(p2) G(p3) [A(p4) skipped] D(p5) G(p6) V(p7) G(p8) K(p9)
# So cumcount within group: 1=V(p1), 2=V(p2), 3=G(p3), 4=D(p5), 5=G(p6), 6=V(p7), 7=G(p8), 8=K(p9)
UPWARD_WT_AAS = {'D', 'V', 'K'}  # p5, p7, p9
```

```python
# ── 1. Load AF2 ipSAE scores ──────────────────────────────────────────────────
print("Loading AF2 ipSAE scores...")
af2_df = pd.read_csv(AF2_PATH)
af2_df = af2_df.rename(columns={'label': 'design'})
af2_cols = ['design', 'ipsae_binder_peptide', 'pae_interaction', 'plddt_binder',
            'plddt_target', 'binder_aligned_rmsd', 'target_aligned_rmsd',
            'mean_pae_bp', 'n_contacts']
af2_df = af2_df[af2_cols].copy()
af2_df = af2_df[af2_df['ipsae_binder_peptide'] >= 0.8]
print(f"  {len(af2_df)} rows")

# ── 2. Load CMS scores ────────────────────────────────────────────────────────
print("Loading CMS scores...")
cms_df = pd.read_csv(CMS_PATH, sep='\t')
cms_df = cms_df.rename(columns={'description': 'design'})
cms_df = cms_df[['design', 'cms_p5_neoepitope', 'cms_peptide_total']].copy()
print(f"  {len(cms_df)} rows")

# ── 3. Load pMHC fold specificity scores ─────────────────────────────────────
print("Loading pMHC fold specificity scores...")
pmhc_df = pd.read_csv(PMHC_PATH, sep='\t')
pmhc_df = pmhc_df.rename(columns={'base_name': 'design'})
pmhc_cols = [
    'design',
    'ppi_pae_int_peptide_on', 'ppi_pae_int_peptide_off',
    'delta_pae_int_peptide',
    'plddt_peptide_on', 'plddt_peptide_off',
    'ppi_pae_int_on', 'ppi_pae_int_off',
    'delta_pae_int'
]
pmhc_df = pmhc_df[pmhc_cols].copy()
print(f"  {len(pmhc_df)} rows")

# ── 4. Load and process MPNN alanine scan ─────────────────────────────────────
print("Loading MPNN alanine scan scores...")
mpnn_raw = pd.read_csv(MPNN_PATH)
mpnn_raw = mpnn_raw.rename(columns={'wt_pdb': 'design'})

# Add per-group cumcount to identify peptide position (1-8, skipping p4=A)
mpnn_raw['pos_idx'] = mpnn_raw.groupby('design').cumcount() + 1
# pos_idx: 1=V(p1), 2=V(p2), 3=G(p3), 4=D(p5), 5=G(p6), 6=V(p7), 7=G(p8), 8=K(p9)

# Verify D->A is at pos_idx=4 for a few designs
sample_check = mpnn_raw[mpnn_raw['pos_idx'] == 4][['design', 'mutant']].head(3)
print(f"  Spot check pos_idx=4 (should be D->A):\n{sample_check.to_string(index=False)}")

# D->A score (p5 neoepitope): pos_idx == 4
mpnn_p5 = mpnn_raw[mpnn_raw['pos_idx'] == 4][['design', 'log_p_mt__minus__log_p_wt']].copy()
mpnn_p5 = mpnn_p5.rename(columns={'log_p_mt__minus__log_p_wt': 'mpnn_delta_p5_DA'})

# Upward-facing mean: pos_idx in {4, 6, 8} = p5(D), p7(V), p9(K)
mpnn_upward = mpnn_raw[mpnn_raw['pos_idx'].isin([4, 6, 8])].groupby('design')['log_p_mt__minus__log_p_wt'].mean().reset_index()
mpnn_upward = mpnn_upward.rename(columns={'log_p_mt__minus__log_p_wt': 'mpnn_delta_upward_mean'})

# Mean across all 8 positions
mpnn_all = mpnn_raw.groupby('design')['log_p_mt__minus__log_p_wt'].mean().reset_index()
mpnn_all = mpnn_all.rename(columns={'log_p_mt__minus__log_p_wt': 'mpnn_delta_all_mean'})

# Merge MPNN metrics
mpnn_df = mpnn_p5.merge(mpnn_upward, on='design').merge(mpnn_all, on='design')
print(f"  {len(mpnn_df)} designs with MPNN scores")

# ── 5. Merge all scores ───────────────────────────────────────────────────────
print("\nMerging all score tables...")

merged = af2_df.copy()
merged = merged.merge(cms_df,   on='design', how='left')
merged = merged.merge(pmhc_df,  on='design', how='left')
merged = merged.merge(mpnn_df,  on='design', how='left')

print(f"  Final merged table: {len(merged)} designs")
print(f"  Missing CMS:   {merged['cms_p5_neoepitope'].isna().sum()}")
print(f"  Missing pMHC:  {merged['delta_pae_int_peptide'].isna().sum()}")
print(f"  Missing MPNN:  {merged['mpnn_delta_p5_DA'].isna().sum()}")
```

```python
# ── 6. Compute per-metric ranks ───────────────────────────────────────────────
# For each metric, rank such that rank 1 = best design

# ipSAE: higher = better
merged['rank_ipsae'] = merged['ipsae_binder_peptide'].rank(ascending=False)

# CMS p5: higher = better (more contact with neoepitope)
merged['rank_cms_p5'] = merged['cms_p5_neoepitope'].rank(ascending=False)

# CMS total: higher = better
merged['rank_cms_total'] = merged['cms_peptide_total'].rank(ascending=False)

# pMHC fold Δ pAE_int_peptide: more negative = better
merged['rank_pmhc_spec'] = merged['delta_pae_int_peptide'].rank(ascending=True)

# pMHC fold on-target pAE_int_peptide: lower = better
merged['rank_pmhc_on'] = merged['ppi_pae_int_peptide_on'].rank(ascending=True)

# MPNN Δ p5 (D->A): more negative = better
merged['rank_mpnn_p5'] = merged['mpnn_delta_p5_DA'].rank(ascending=True)

# MPNN upward-facing mean: more negative = better
merged['rank_mpnn_upward'] = merged['mpnn_delta_upward_mean'].rank(ascending=True)

# # ── 7. Combined rank ──────────────────────────────────────────────────────────
# # Primary filters already applied (ipSAE >= 0.8) — all 199 pass
# # Combine ranks with equal weighting across the three specificity metrics:
# #   pMHC fold Δ pAE, CMS p5, MPNN Δ p5
# # Plus ipSAE as a quality anchor
# merged['rank_combined'] = (
#     merged['rank_ipsae'] +
#     merged['rank_cms_p5'] +
#     merged['rank_pmhc_spec'] +
#     merged['rank_mpnn_p5']
# ) / 4

# merged = merged.sort_values('rank_combined')

# ── 8. Save ───────────────────────────────────────────────────────────────────
out_path = os.path.join(OUT_DIR, 'merged_scores_ranked.tsv')
merged.to_csv(out_path, sep='\t', index=False)
print(f"\nSaved merged ranked table to {out_path}")

# ── 9. Summary table (top 30) ─────────────────────────────────────────────────
summary_cols = [
    'design',
    'ipsae_binder_peptide',
    'pae_interaction',
    'plddt_binder',
    'cms_p5_neoepitope',
    'cms_peptide_total',
    'delta_pae_int_peptide',
    'ppi_pae_int_peptide_on',
    'ppi_pae_int_peptide_off',
    'mpnn_delta_p5_DA',
    'mpnn_delta_upward_mean',
    'mpnn_delta_all_mean',
    'rank_ipsae',
    'rank_cms_p5',
    'rank_pmhc_spec',
    'rank_mpnn_p5',
#     'rank_combined',
]
```

```python
# ── 10. Correlation between metrics ──────────────────────────────────────────
print("\n── Spearman correlations between specificity metrics ──")
spec_cols = [
    'ipsae_binder_peptide',
    'cms_p5_neoepitope',
    'cms_peptide_total',
    'delta_pae_int_peptide',
    'ppi_pae_int_peptide_on',
    'mpnn_delta_p5_DA',
    'mpnn_delta_upward_mean',
    'mpnn_delta_all_mean',
]
# Friendlier axis labels
labels = [
    'ipSAE\n(binder-peptide)',
    'CMS p5\n(neoepitope)',
    'CMS\n(peptide total)',
    'Δ pAE_int_peptide\n(pMHC fold spec.)',
    'pAE_int_peptide\n(on-target)',
    'MPNN Δ log P\n(p5, D→A)',
    'MPNN Δ log P\n(upward mean)',
    'MPNN Δ log P\n(all positions)',
]
corr = merged[spec_cols].corr(method='spearman').round(3)
print(corr.to_string())

corr_path = os.path.join(OUT_DIR, 'metric_correlations.tsv')
corr.to_csv(corr_path, sep='\t')
print(f"Correlation matrix saved to {corr_path}")

print("\nDone.")
```

```python
fig, ax = plt.subplots(figsize=(9, 7))

im = ax.imshow(corr.values, cmap='RdBu_r', vmin=-1, vmax=1, aspect='auto')

# Colorbar
cbar = plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
cbar.set_label("Spearman ρ", fontsize=11)
cbar.set_ticks([-1, -0.5, 0, 0.5, 1])

# Tick labels
ax.set_xticks(range(len(labels)))
ax.set_yticks(range(len(labels)))
ax.set_xticklabels(labels, fontsize=8, rotation=45, ha='right', rotation_mode='anchor')
ax.set_yticklabels(labels, fontsize=8)

# Annotate cells with correlation values
for i in range(len(spec_cols)):
    for j in range(len(spec_cols)):
        val = corr.values[i, j]
        # Use white text on dark cells, black on light cells
        text_color = 'white' if abs(val) > 0.6 else 'black'
        ax.text(j, i, f'{val:.2f}', ha='center', va='center',
                fontsize=7.5, color=text_color, fontweight='bold')

ax.set_title('Spearman Correlation Between Scoring Metrics\n(199 designs, Round 2)',
             fontsize=12, pad=15)

plt.tight_layout()
plt.show()
# fig.savefig(f"{OUT_DIR}/metric_correlation_heatmap.png", dpi=150, bbox_inches='tight')
# print(f"Saved to {OUT_DIR}/metric_correlation_heatmap.png")
# plt.close()
```
